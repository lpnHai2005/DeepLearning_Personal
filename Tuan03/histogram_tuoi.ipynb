```python
import matplotlib.pyplot as plt

# Dữ liệu tuổi hợp lệ
tuoi = [25, 30, 22, 40, 35, 29, 50, 27, 31]

# Chia nhóm theo yêu cầu
nhom = ["20-25", "26-30", "31-35", "36-40", "41-50"]
so_luong = [2, 3, 2, 1, 1]

plt.figure(figsize=(8, 5))

plt.bar(
    nhom,
    so_luong,
    edgecolor="black"
)

plt.title("Histogram phân phối Tuổi")
plt.xlabel("Nhóm tuổi")
plt.ylabel("Số lượng người")

plt.grid(axis="y", alpha=0.3)

# Hiển thị số lượng trên đầu mỗi cột
for i, value in enumerate(so_luong):
    plt.text(i, value + 0.05, str(value), ha="center")

plt.tight_layout()

# Lưu ảnh
plt.savefig("histogram_tuoi.png", dpi=300, bbox_inches="tight")

plt.show()
```
