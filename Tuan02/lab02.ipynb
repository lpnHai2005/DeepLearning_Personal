## So sánh `torch.rand()` và `torch.randn()`

| Đặc điểm | `torch.rand()` | `torch.randn()` |
|---|---|---|
| **Phân phối** | Uniform (đều) | Normal (chuẩn) |
| **Khoảng giá trị** | [0, 1) | (-∞, +∞) |
| **Mean** | 0.5 | 0 |
| **Std** | ≈ 0.289 | 1 |
| **Ứng dụng** | Khởi tạo trọng số, xác suất | Tạo nhiễu Gaussian, sampling |

Tóm lại:

torch.rand() sinh giá trị ngẫu nhiên đều trong [0, 1) → luôn dương, không có giá trị âm.

torch.randn() sinh giá trị ngẫu nhiên theo phân phối chuẩn (mean=0, std=1) → có cả giá trị âm và dương.

## Bài 7. Các hàm thống kê trên Tensor

Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
print("\n===== dim = 1 =====")

print("\nMean:")
print(torch.mean(a, dim=1))

print("\nStandard deviation:")
print(torch.std(a, dim=1))

print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

Nhận xét:
- Với dim=0, các phép toán được thực hiện theo từng cột. Mean và Standard deviation trả về Tensor có 4 phần tử, tương ứng với 4 cột. Cumulative sum tính tổng tích lũy theo chiều dọc.
- Với dim=1, các phép toán được thực hiện theo từng hàng. Mean và Standard deviation trả về Tensor có 3 phần tử, tương ứng với 3 hàng. Cumulative sum tính tổng tích lũy theo chiều ngang.
- Như vậy, dim=0 thực hiện tính toán theo chiều cột, còn dim=1 thực hiện tính toán theo chiều hàng.

## Bài 8. Autograd cơ bản

Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

manual_grad = 2 * 3.6

print("Đạo hàm thủ công =", manual_grad)
print("Autograd =", x.grad.item())

x = torch.tensor(3.6, requires_grad=True)

y = 3 * x**2 + 2 * x + 1

y.backward()

print("\n=== y = 3*x² + 2*x + 1 ===")
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

manual_grad = 6 * 3.6 + 2

print("Đạo hàm thủ công =", manual_grad)
print("Autograd =", x.grad.item())
```

## Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)

print("\nTarget coefficients:")
print(torch.tensor([[1.0], [2.0], [3.0]]))


print("\nComparison:")
print("w1 (x^2):", w[0].item(), "so với 1")
print("w2 (x):  ", w[1].item(), "so với 2")
print("w3:      ", w[2].item(), "so với 3")
```

Nhận xét:
- Ba hệ số thực tế của đa thức là [1, 2, 3].
- Sau quá trình huấn luyện bằng NAdam, ba hệ số học được xấp xỉ [1, 2, 3].
- Sai số giữa hệ số học được và hệ số thực tế nhỏ, cho thấy mô hình đã học được gần đúng đa thức ban đầu.

Giải thích các lệnh:

- optimizer.zero_grad():
Xóa gradient của lần lặp trước để tránh gradient bị cộng dồn.

- mse.backward():
Tính gradient của hàm mất mát MSE đối với các hệ số w bằng Autograd.

- optimizer.step():
NAdam sử dụng gradient vừa tính để cập nhật các hệ số w, giúp giảm giá trị MSE.

## Bài 10. Giải hệ phương trình bằng Autograd

Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

print("\nKiểm tra các phương trình:")

print("A + B =", (A + B).item(), "-> yêu cầu = 9")
print("C - D =", (C - D).item(), "-> yêu cầu = 1")
print("A + C =", (A + C).item(), "-> yêu cầu = 8")
print("B - D =", (B - D).item(), "-> yêu cầu = 2")

print("\nSai số:")
print("y1 =", (A + B - 9).item())
print("y2 =", (C - D - 1).item())
print("y3 =", (A + C - 8).item())
print("y4 =", (B - D - 2).item())
```

Nhận xét:
- Sau khi tối ưu, các giá trị A, B, C, D tìm được thỏa mãn gần đúng cả bốn phương trình.
- Sai số của bốn phương trình tiến gần về 0.
- Tuy nhiên, bài toán có nhiều nghiệm vì bốn phương trình không độc lập hoàn toàn.
- Đặt D = t thì:

    B = t + 2

    C = t + 1
    
    A = 7 - t
- Với mọi giá trị t, bốn phương trình đều được thỏa mãn.
- Vì vậy, hệ phương trình có vô số nghiệm thay vì chỉ có một nghiệm duy nhất.

4. Thay đa thức trong Bài 9 bằng y = 2x² - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
polynomial = np.poly1d([2, -3, 5])

N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)
print("Hệ số ban đầu:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()

    y_pred = x @ w

    mse = torch.mean(torch.square(y - y_pred))

    mse.backward()

    optimizer.step()

print("\nHệ số học được:")
print(w)

print("\nHệ số thực tế:")
print(torch.tensor([[2.0], [-3.0], [5.0]]))

print("\nSo sánh:")
print("Hệ số x²:", w[0].item(), "-> thực tế: 2")
print("Hệ số x :", w[1].item(), "-> thực tế: -3")
print("Hệ số tự do:", w[2].item(), "-> thực tế: 5")
```
